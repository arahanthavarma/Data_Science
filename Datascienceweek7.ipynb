```python
numeric_cols = df.select_dtypes(include="number").columns

df[numeric_cols] = df[numeric_cols].fillna(
    df[numeric_cols].median()
)

df.dropna(inplace=True)

print(df.isnull().sum())
```

```python
print("Duplicates:", df.duplicated().sum())

df.drop_duplicates(inplace=True)

print("After removing duplicates:", df.duplicated().sum())
```

```python
from sklearn.preprocessing import OrdinalEncoder

order = [["Whole Year", "Kharif", "Rabi", "Summer", "Winter"]]

oe = OrdinalEncoder(
    categories=order,
    handle_unknown="use_encoded_value",
    unknown_value=-1
)

df["Season_Encoded"] = oe.fit_transform(df[["Season"]])

print(df[["Season", "Season_Encoded"]].head())
```

```python
data = df["Production"].dropna()

print("Mean:", data.mean())
print("Median:", data.median())
print("Mode:", data.mode().tolist())
```

```python
print("Standard Deviation:", data.std())
print("Variance:", data.var())
```

```python
minimum = data.min()
maximum = data.max()
value_range = maximum - minimum

print("Minimum:", minimum)
print("Maximum:", maximum)
print("Range:", value_range)
```

```python
Q1 = data.quantile(0.25)
Q2 = data.quantile(0.50)
Q3 = data.quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q2:", Q2)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
print("Skewness:", data.skew())
```

```python
print("Kurtosis:", data.kurt())
```

```python
Q1 = data.quantile(0.25)
Q3 = data.quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = data[
    (data < lower_limit) | (data > upper_limit)
]

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Number of Outliers:", len(outliers))
print(outliers)
```

```python
mean_value = data.mean()
std_value = data.std()

if mean_value != 0:
    cv = (std_value / abs(mean_value)) * 100
    print("Coefficient of Variation:", cv, "%")
else:
    print("CV cannot be calculated because mean is zero.")
```
